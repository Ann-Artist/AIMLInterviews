感知机算法是一种线性分类算法，用于将数据分为两个类别。它是一种简单的算法，会在分类过程中根据犯下的错误进行学习，并调整输入特征的权重，以提高分类准确率。 

```python 
y_pred = sign(w0 + w1*x1 + w2*x2 + ... + wn*xn)
wi = wi + learning_rate * (target - y_pred) * xi
```

下面是一个用Python实现的感知机算法：

````python
import numpy as np

class Perceptron:
    def __init__(self, lr=0.01, n_iter=100):
        self.lr = lr
        self.n_iter = n_iter

    def fit(self, X, y):
        self.weights = np.zeros(1 + X.shape[1])
        self.errors = []

        for _ in range(self.n_iter):
            errors = 0
            for xi, target in zip(X, y):
                update = self.lr * (target - self.predict(xi))
                self.weights[1:] += update * xi
                self.weights[0] += update
                errors += int(update != 0.0)
            self.errors.append(errors)
        return self

    def net_input(self, X):
        return np.dot(X, self.weights[1:]) + self.weights[0]

    def predict(self, X):
        return np.where(self.net_input(X) >= 0.0, 1, -1)
````

`Perceptron` 类包含以下方法：

`__init__(self, lr=0.01, n_iter=100)`：使用学习率（`lr`）和训练迭代次数（`n_iter`）初始化感知机。

`fit(self, X, y)`：在输入数据 `X` 和目标标签 `y` 上训练感知机。该方法会先将权重初始化为 0，然后对数据迭代 `n_iter` 次，并在每次误分类后调整权重。该方法返回训练后的感知机。

`net_input(self, X)`：计算输入与偏置的加权和。

`predict(self, X)`：基于当前权重预测给定输入 `X` 的类别标签。

要使用感知机算法，你可以先创建一个 `Perceptron` 类的实例，然后调用 `fit` 方法并传入输入数据 `X` 和目标标签 `y`。下面是一个使用示例：

````python
X = np.array([[2.0, 1.0], [3.0, 4.0], [4.0, 2.0], [3.0, 1.0]])
y = np.array([-1, 1, 1, -1])
perceptron = Perceptron()
perceptron.fit(X, y)

new_X = np.array([[5.0, 2.0], [1.0, 3.0]])
perceptron.predict(new_X)

````
